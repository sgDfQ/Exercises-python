```python
import pandas as pd
import seaborn as sns
import ssl

ssl._create_default_https_context = ssl._create_unverified_context
df = sns.load_dataset("titanic")
print(df.head())
```

```python
# 1. Combien de passagers avaient moins de 18 ans ? #
df[df['age'] < 18].shape[0]
resultat_moins_18_ans = df[df['age'] < 18].shape[0]
print(f"Moins de 18 ans : {resultat_moins_18_ans}")
```

```python
# 2. Afficher les 5 billets les plus chers avec la classe et le port d'embarquement. #
billets_plus_chers = df.nlargest(5, 'fare')
print (billets_plus_chers[['fare', 'class', 'embark_town']])
```

```python
# 3. Quelle part des passagers de 1ere classe sont de femmes ? #
passagers_premiere_classe = df[df['class'] == 'First']
femmes_premiere_classe = passagers_premiere_classe[passagers_premiere_classe['sex'] == 'female']
part_femmes_premiere_classe = len(femmes_premiere_classe) / len(passagers_premiere_classe) * 100
print(f"Part des femmes en 1ère classe : {part_femmes_premiere_classe:.2f}%")
```

```python
# 4. Quel est le taux de survie des femmes de 1er classe ?

taux_survie_femmes_premiere_classe = df[(df['survived'] == 1) & (df['class'] == 'First') & (df['sex'] == 'female')]
femmes_premiere_classe = passagers_premiere_classe[passagers_premiere_classe['sex'] == 'female']
print(f"Taux de survie des femmes de 1ère classe : {len(taux_survie_femmes_premiere_classe) / len(femmes_premiere_classe) * 100:.2f}%")

# 4.1 Et des hommes de 3e classe ? 
hommes_troisieme_classe = df[(df['class'] == 'Third') & (df['sex'] == 'male')]
hommes_survivants = hommes_troisieme_classe[hommes_troisieme_classe['survived'] == 1]
print(f"Taux de survie des hommes de 3e classe : {len(hommes_survivants) / len(hommes_troisieme_classe) * 100:.2f}%")
```
