This part is about doing grafphs with matplotlib and seaborn. You will need to use the data from the previous exercises.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import ssl

ssl._create_default_https_context = ssl._create_unverified_context
titanic_propre = pd.read_csv('../exercice-08/data/titanic_propre.csv')
taxis = sns.load_dataset("taxis")
```

1. Quel âge avaient les passagers ?

```python
sns.histplot(data=titanic_propre, x='age')
plt.title("Âge des passagers")
plt.xlabel("Âge")
plt.ylabel("Nombre de passagers")
plt.show()
```

2. La survie selon la classe et le sexe

```python
sns.barplot(data=titanic_propre, x='pclass', y='survived', hue='sex')
plt.title("Survie selon la classe et le sexe")
plt.xlabel("Classe")
plt.ylabel("Taux de survie")
plt.show()
```

3. Le prix du billet selon la classe

```python
sns.boxplot(data=titanic_propre, x='pclass', y='fare')
plt.title("Prix du billet selon la classe")
plt.xlabel("Classe")
plt.ylabel("Prix du billet")
plt.show()
```

4. Taxis : le chiffre d'affaires jour après jour

```python
taxis['jour'] = taxis['pickup'].dt.date
chiffre_affaires_par_jour = taxis.groupby('jour')['total'].sum()

plt.plot(chiffre_affaires_par_jour.index, chiffre_affaires_par_jour.values)
plt.title("Chiffre d'affaires des taxis par jour")
plt.xlabel("Jour")
plt.ylabel("Chiffre d'affaires ($)")
plt.xticks(rotation=45)
plt.show()
```

5. Taxis : le lien entre distance et prix

```python
sns.scatterplot(data=taxis, x='distance', y='fare')
plt.title("Lien entre distance et prix")
plt.xlabel("Distance (miles)")
plt.ylabel("Prix ($)")
plt.show()
```
