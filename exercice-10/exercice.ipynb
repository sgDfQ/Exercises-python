```python
import pandas as pd

titanic_propre = pd.read_csv('../exercice-08/data/titanic_propre.csv')
```

1. Créer la table ports de la diapositive précédente

```python
ports = pd.DataFrame({
    'embark_town': ['Southampton', 'Cherbourg', 'Queenstown'],
    'pays': ['Angleterre', 'France', 'Irlande']
})
print(ports)
```

2. La joindre à titanic_propre en how="left" et vérifier le nombre de lignes

```python
jointure = titanic_propre.merge(ports, on='embark_town', how='left')

# Le nombre de lignes doit rester le même
print(f"Avant : {len(titanic_propre)} lignes, après : {len(jointure)} lignes")
```

3. Compter les passagers et calculer le taux de survie par pays

```python
# Nombre de passagers par pays
passagers_par_pays = jointure.groupby('pays')['survived'].count()
print(passagers_par_pays)

# Taux de survie par pays (moyenne de survived x 100)
taux_survie_par_pays = jointure.groupby('pays')['survived'].mean() * 100
print(taux_survie_par_pays)
```

4. Retirer Queenstown de ports, refaire la jointure en "inner" puis en "left" : que se passe-t-il ? L'écrire en Markdown

```python
ports_sans_queenstown = ports[ports['embark_town'] != 'Queenstown']

jointure_inner = titanic_propre.merge(ports_sans_queenstown, on='embark_town', how='inner')
print(f"inner : {len(jointure_inner)} lignes")

jointure_left = titanic_propre.merge(ports_sans_queenstown, on='embark_town', how='left')
print(f"left : {len(jointure_left)} lignes")
print(f"Pays vides (NaN) : {jointure_left['pays'].isna().sum()}")
```

Que se passe-t-il ?

- **inner** : 814 lignes. Les 77 passagers partis de Queenstown disparaissent, car leur port n'est plus dans la table ports. On perd des données.
- **left** : 891 lignes. Tous les passagers sont gardés, mais les 77 passagers de Queenstown ont un pays vide (NaN).
