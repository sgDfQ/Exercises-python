```python
import pandas as pd

titanic_propre = pd.read_csv('../exercice-08/data/titanic_propre.csv')
```

1. Les femmes ont-elles mieux survécu que les hommes dans chaque classe ?

```python
taux_survie_classe_sexe = titanic_propre.groupby(['pclass', 'sex'])['survived'].mean() * 100
print(taux_survie_classe_sexe)
```

Oui : dans chaque classe, les femmes ont beaucoup mieux survécu que les hommes (1ère : 96.8 % contre 36.9 %, 2e : 92.1 % contre 15.7 %, 3e : 50.0 % contre 13.5 %).

2. Les enfants ont-ils été prioritaires ? Taux de survie par tranche d'âge

```python

titanic_propre['tranche_age'] = pd.cut(titanic_propre['age'], bins=[0, 12, 18, 60, 100], labels=['Enfant', 'Ado', 'Adulte', 'Senior'])

taux_survie_par_age = titanic_propre.groupby('tranche_age', observed=True)['survived'].mean() * 100
print(taux_survie_par_age)
```

Oui : les enfants ont le meilleur taux de survie (58.0 %), puis les ados (42.9 %), les adultes (36.6 %) et les seniors (22.7 %).

3. Quel port a le prix moyen le plus élevé ? La répartition des classes l'explique-t-elle ?

```python
prix_moyen_par_port = titanic_propre.groupby('embark_town')['fare'].mean()
print(prix_moyen_par_port)


classes_par_port = pd.crosstab(titanic_propre['embark_town'], titanic_propre['pclass'], normalize='index')
print(classes_par_port)
```

Cherbourg a le prix moyen le plus élevé (59.95 contre 27.24 pour Southampton et 13.28 pour Queenstown). Oui, les classes l'expliquent : 51 % des passagers de Cherbourg sont en 1ère classe, alors que 94 % de ceux de Queenstown sont en 3e classe.
