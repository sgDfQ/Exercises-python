```python
import pandas as pd
import seaborn as sns
import ssl
import os

ssl._create_default_https_context = ssl._create_unverified_context
df = sns.load_dataset("titanic")
```

1. Traiter les manquants : age par la médiane de sa classe et de son sexe,
embark_town, deck

```python
mediane_age = df.groupby(['class', 'sex'])['age'].transform('median')
df['age'] = df['age'].fillna(mediane_age)

df['embark_town'] = df['embark_town'].fillna('Southampton')

df = df.drop(columns=['deck'])
```

2. Examiner les doublons et décider s'il faut les supprimer

```python
nombre_doublons = df.duplicated().sum()
print(f"Doublons : {nombre_doublons}")
```

3. Supprimer les colonnes redondantes, corriger les types

```python
df = df.drop(columns=['embarked', 'class', 'alive', 'who', 'adult_male', 'alone'])
df['sex'] = df['sex'].astype('category')
df['embark_town'] = df['embark_town'].astype('category')
```

4. Repérer les billets hors bornes et décider

```python
billets_gratuits = (df['fare'] == 0).sum()
billets_tres_chers = (df['fare'] > 500).sum()
print(f"Billets à 0 : {billets_gratuits}, billets > 500 : {billets_tres_chers}")
```

5. Enregistrer dans data/titanic_propre.csv avec to_csv(..., index=False)

```python
os.makedirs('data', exist_ok=True)
df.to_csv('data/titanic_propre.csv', index=False)
```
